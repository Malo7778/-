```python
                pending = []
            current = key
            pending.append(row)
    if pending:
        result = summarize(pending)
        if result is not None:
            yield result


def subject_epochs(sid, revision):
    if not sid.startswith('S') or not sid[1:].isdigit():
        raise ValueError('Subject ID should look like S002.')
    url = f'https://huggingface.co/datasets/bsaenz/dreamt/resolve/{revision}/data_64Hz/{sid}_whole_df.csv'
    # Use the CSV builder to exclude 100 Hz PSG and participant metadata files.
    ds = load_dataset('csv', data_files={'train': url}, split='train', streaming=True)
    return epochs(ds.iter(batch_size=8192))


def train(args):
    if len(args.subjects) < 4 or len(set(args.subjects)) != len(args.subjects):
        raise ValueError('Provide at least four distinct subjects; last two are held out.')
    X, y, groups = [], [], []
    for sid in args.subjects:
        count = 0
        for features, label, _ in subject_epochs(sid, args.revision):
            X.append(features)
            y.append(label)
            groups.append(sid)
            count += 1
        if not count:
            raise ValueError(f'{sid}: no valid labeled epochs.')
        print(f'{sid}: {count} epochs', flush=True)
    X, y, groups = np.array(X), np.array(y), np.array(groups)
    test = np.isin(groups, args.subjects[-2:])
    if len(set(y[~test])) < 2:
        raise ValueError('Training subjects contain fewer than two sleep stages.')
    model = make_pipeline(SimpleImputer(strategy='median', keep_empty_features=True),
                          RandomForestClassifier(n_estimators=200, min_samples_leaf=5,
                                                 class_weight='balanced', n_jobs=-1, random_state=42))
    model.fit(X[~test], y[~test])
    predicted = model.predict(X[test])
    metrics = {
        'train_subjects': args.subjects[:-2], 'test_subjects': args.subjects[-2:],
        'revision': args.revision, 'training_epochs': int((~test).sum()), 'test_epochs': int(test.sum()),
        'balanced_accuracy': balanced_accuracy_score(y[test], predicted),
        'classification_report': classification_report(y[test], predicted, labels=STAGES, output_dict=True, zero_division=0),
        'confusion_matrix_labels': STAGES,
        'confusion_matrix': confusion_matrix(y[test], predicted, labels=STAGES).tolist(),
        'missing_training_stages': sorted(set(STAGES) - set(y[~test])),
        'limitation': 'Small subject-held-out baseline; not clinical validation or evidence of sleep improvement.'
    }
    out = Path(args.output)
    out.mkdir(parents=True, exist_ok=True)
    joblib.dump({'model': model, 'channels': CHANNELS, 'revision': args.revision}, out / 'model.joblib')
    (out / 'metrics.json').write_text(json.dumps(metrics, ensure_ascii=False, indent=2), encoding='utf-8')
    print(json.dumps(metrics, ensure_ascii=False, indent=2))


def predict(args):
    # Only load your own trusted model files: joblib can execute serialized code.
    bundle = joblib.load(args.model)
    if bundle['channels'] != CHANNELS:
        raise ValueError('Model channel schema does not match.')
    rows = list(epochs(pd.read_csv(args.csv, chunksize=8192), labeled=False))
    if not rows:
        raise ValueError('No complete 30-second 64 Hz epochs found.')
    X = np.array([r[0] for r in rows])
    model = bundle['model']
    labels, confidence = model.predict(X), model.predict_proba(X).max(axis=1)
    table = pd.DataFrame({'start_second': [r[2] for r in rows], 'predicted_stage': labels,
                          'model_vote_fraction': confidence})
    out = Path(args.output)
    out.mkdir(parents=True, exist_ok=True)
    table.to_csv(out / 'predictions.csv', index=False)
    sleep_minutes = int(np.isin(labels, ['N1', 'N2', 'N3', 'R']).sum()) / 2
    distribution = {stage: int((labels == stage).sum()) / 2 for stage in STAGES}
    report = [
        '# 睡眠分析研究报告', '',
        f'有效分析时长：{len(rows) / 2:.1f} 分钟；预测睡眠时长：{sleep_minutes:.1f} 分钟。',
        f'各阶段预测分钟数：{distribution}', '',
        '以上仅覆盖通过质量检查的时间窗；排除的窗口或数据间隙不计入统计。',
        '这是实验模型预测，不是医学测量。模型投票比例不等于经过校准的准确概率。',
        '数据来源设备、单位或人群变化可能导致明显误差。不要把普通手表导出数据直接当成兼容输入。', '',
        '## 生活习惯记录提醒（固定规则，未从 DREAMT 学得）', '',
        '可以记录每日作息、睡前活动和第二天的主观感受，观察自己的变化。',
        '若睡眠困扰持续或影响日常生活，可将记录带给专业人员讨论。',
        '该数据集没有干预效果标签，因此本程序不能证明这些提醒能改善睡眠。'
    ]
    (out / 'report.md').write_text('\n'.join(report), encoding='utf-8')
    print(f'Report saved: {out.resolve()}')


def main():
    parser = argparse.ArgumentParser(description=__doc__)
    commands = parser.add_subparsers(dest='command', required=True)
    training = commands.add_parser('train')
    training.add_argument('--subjects', nargs='+', default=['S002', 'S003', 'S004', 'S005', 'S006', 'S007'])
    training.add_argument('--revision', default=REVISION)
    training.add_argument('--output', default='artifacts')
    training.set_defaults(run=train)
    prediction = commands.add_parser('predict')
    prediction.add_argument('--csv', required=True)
    prediction.add_argument('--model', default='artifacts/model.joblib')
    prediction.add_argument('--output', default='report')
    prediction.set_defaults(run=predict)
    args = parser.parse_args()
    args.run(args)


if __name__ == '__main__':
    main()
```
